Code for subtract function

```python
def add(a,b):
    return (a+b)

# Testing code
print(add(10, 5),
add(20, 8),
add(15, 3))

def sum_list(list):
    sum = list[0]
    for number in list[1:]:
        sum = add(sum, number)
    return sum

print(sum_list([1, 2, 3, 4, 5]))
print(sum_list([10, 20, 30, 40, 50]))
print(sum_list([100, 200, 300, 400, 500]))
```
