```python
#Multiply function
def multiply(a, b):
    return a * b
#square function
def square(x):
    return multiply(x, x)
```

```python
#Multiply two numbers
#Inputs for user's choice
a=int(input('first number'))
b=int(input('second number'))
#print the final result
print('the multiplication of ', a, 'and', b ,'is' ,multiply(a,b))
```

```python
#Square function
#Inputs for user's choice
x=int(input('number to square'))
#Print the final result
print('the square of ', x, 'is' ,square(x))
```
