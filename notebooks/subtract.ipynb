Code for subtract function

```python
def substract(initial, diff):
    return (initial - diff)

# Testing code
print(substract(10, 5),
substract(20, 8),
substract(15, 3))

def add(a,b):
    return (a+b)

```

```python
#this code substracts two numbers from one in the order a - b - c, this is done by first adding b +c and substracting from a using the substract and add functions.
a=int(input('first number'))
b=int(input('second number'))
c=int(input('third number'))
ans=substract(a,add(b,c))
print(ans)
```
